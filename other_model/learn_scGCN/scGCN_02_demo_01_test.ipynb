# run scGCN

````python
from pathlib import Path
import scanpy as sc
from scGCN__init__ import scGCN,select_feature,scGCN_train
from scGCN__init__ import adata_norm_scale_for_scGCN,scGCN_get_res

p_data = Path('process_data')
adata_1, adata_2 = sc.read_h5ad(p_data.joinpath('temp_raw_1.h5ad')), \
    sc.read_h5ad(p_data.joinpath('temp_raw_2.h5ad'))

key_label = 'type'

p_out = Path('scGCN_Demo')
p_out.mkdir(exist_ok=True,parents=True)
features = select_feature(adata_1, key_label)
adata_1, adata_2 = adata_1[:, features].copy(), adata_2[:, features].copy()

adata_1.obs.to_csv(p_out.joinpath('obs_ref.csv'),index=True)
adata_2.obs.to_csv(p_out.joinpath('obs_que.csv'),index=True)

# 将adata_1 adata_2 的副本标准化后传入graph_construct
scGCN.graph.graph_construct(adata_norm_scale_for_scGCN(adata_1.copy()),
                            adata_norm_scale_for_scGCN(adata_2.copy()),
                            'type', p_out)

# 调用data.input_data
scGCN.data.input_data(p_out, adata_1, adata_2, key_label)

# epochs = 20 for test
scGCN_train(p_out, p_out, epochs=200)
````

````python
dsn1 = '_m'
dsn2 = '_h'
key_label = 'type'

df_scGCN_res = scGCN_get_res(p_out,key_label,key_label,dsn1,dsn2)
df_scGCN_res['UMAP1'] = 1
df_scGCN_res['UMAP2'] = 1
df_scGCN_res = df_scGCN_res.loc[:,
    'UMAP1,UMAP2,dataset,cell_type,true_label,pre_label,max_prob,is_right'.split(',')]
df_scGCN_res.to_csv(p_out.joinpath('obs.csv'),index=True)
````
